# Word Document Processing

Word documents are another common source of data in real RAG projects.

In this notebook, I’m learning how to load `.docx` files and see how different loaders extract the content.

I’m trying two approaches here:

- `Docx2txtLoader`
- `UnstructuredWordDocumentLoader`

The main thing I want to understand is how much information each loader gives me and how the extracted content can be used later in the RAG pipeline.


## Loading the Word Document Loaders

I’m importing two different loaders so I can process the same Word document in two ways.

The first loader mainly focuses on extracting the document text.

The second loader can give me more information about the structure of the document, such as titles, paragraphs, list items, and tables.


```python
from langchain_community.document_loaders import (
    Docx2txtLoader,
    UnstructuredWordDocumentLoader
)
```

## 1. Using Docx2txtLoader

I’m starting with `Docx2txtLoader`.

This is a simple way to extract the text from a Word document.

I want to check:

- How many documents are created
- What the extracted text looks like
- What metadata is available


```python
print("1️⃣ Using Docx2txtLoader")

try:
    docx_loader = Docx2txtLoader(
        "E:/RAG/data/course_samples/word_files/proposal.docx"
    )

    docs = docx_loader.load()

    print(f"✅Loaded {len(docs)} document(s)")
    print(f"Content preview: {docs[0].page_content[:200]}...")
    print(f"Metadata: {docs[0].metadata}")

except Exception as e:
    print(f"Error: {e}")
```

### What I noticed

The Word document is converted into a LangChain `Document`.

The actual text is stored in `page_content`, while basic file information is stored in `metadata`.

This is similar to what I saw with text files and PDFs.


## 2. Using UnstructuredWordDocumentLoader

Now I’m trying `UnstructuredWordDocumentLoader` on the same Word document.

This approach gives me more information about the structure of the document.

Instead of getting only one large piece of text, I can get individual elements such as:

- Titles
- Paragraphs
- List items
- Tables

This can be useful when the structure of the document matters.


```python
print("\n2️⃣ Using UnstructuredWordDocumentLoader")

try:
    unstructured_loader = UnstructuredWordDocumentLoader(
        "E:/RAG/data/course_samples/word_files/proposal.docx",
        mode="elements"
    )

    unstructured_docs = unstructured_loader.load()

    print(f"✅ Loaded {len(unstructured_docs)} elements")

    for i, doc in enumerate(unstructured_docs[:3]):
        print(f"\nElement {i + 1}:")
        print(f"Type: {doc.metadata.get('category', 'unknown')}")
        print(f"Content: {doc.page_content[:100]}...")

except Exception as e:
    print(f"Error: {e}")
```

### Looking at the Extracted Elements

I’m printing the complete list of extracted elements here to understand what the loader is actually giving me.

I’m mainly looking at the relationship between the text and the metadata.


```python
unstructured_docs
```

### What I noticed

This loader gives me much more structured information.

For example, the document can contain different categories such as:

- `Title`
- `NarrativeText`
- `ListItem`
- `Table`

The metadata also contains additional information such as the source file, category, parent element, and other document details.

This can be useful when I want to preserve the original structure of a Word document instead of treating everything as plain text.


## Comparing the Two Approaches

Both loaders can extract content from a Word document, but they give me different levels of information.

`Docx2txtLoader` gives me a simpler text-based representation.

`UnstructuredWordDocumentLoader` can break the document into individual elements and provide more structural metadata.

For a simple document, the first approach may be enough.

For documents where headings, lists, tables, and structure are important, the second approach gives me more information to work with.


```python
print("WORD DOCUMENT LOADER COMPARISON")

print("\nDocx2txtLoader:")
print("  - Simple text extraction")
print("  - Easy to use")
print("  - Useful when document structure is not important")

print("\nUnstructuredWordDocumentLoader:")
print("  - Extracts individual document elements")
print("  - Provides more metadata")
print("  - Can identify titles, text, lists, and tables")
```

## My Takeaway

Word documents can contain more structure than a simple text file.

In this notebook, I learned how to load a `.docx` file using two different loaders and compare the type of information they return.

The main difference I noticed is:

**Docx2txtLoader → Simple text extraction**

**UnstructuredWordDocumentLoader → Structured document elements + metadata**

For a RAG pipeline, the way I parse a document matters because the information I preserve during ingestion can affect the chunking and retrieval stages later.

The flow I’m taking from this notebook is:

**Word Document → Load → Extract Content → Preserve Structure/Metadata → Prepare for Chunking**
