## 6. Understanding RAG in Simple Language

### What Is RAG?

RAG stands for **Retrieval-Augmented Generation**.

Let's understand the three words:

- **Retrieval:** Finding useful information from our documents.
- **Augmented:** Adding that information to the question as extra context.
- **Generation:** Using a language model to generate an answer.

In simple words, RAG first looks for relevant information and then uses that information to answer a question.

## 18. Gemini Embedding Model

### What Am I Doing?

I am connecting Gemini's embedding model to my document-processing pipeline.

### Why Am I Doing This?

My documents have already been loaded and split into chunks. However, ChromaDB needs numerical vectors to perform vector similarity search.

Gemini converts text into vectors that represent semantic information about the text.

### Which Components Am I Using?

- `GeminiEmbeddings`: A small custom adapter that connects Gemini to LangChain's embedding interface.
- `embed_query()`: Converts one text or search query into a vector.
- `embed_documents()`: Converts multiple document chunks into vectors.

### Why Do I Need an Adapter?

I have already tested Gemini's embedding API directly. However, LangChain's ChromaDB integration expects an embedding object with standard methods such as `embed_query()` and `embed_documents()`.

The adapter connects these two interfaces without changing how ChromaDB works.

### What Should I Observe?

A sample sentence should be converted into a numerical vector. I will inspect its dimensions and a few sample values.


````python
# Code: Gemini adapter

class GeminiEmbeddings(Embeddings):
    """Connect Gemini embeddings to LangChain's embedding interface."""

    def __init__(self, gemini_client, model=EMBEDDING_MODEL):
        self.client = gemini_client
        self.model = model

    def embed_documents(self, texts):
        vectors = []

        for text in texts:
            result = self.client.models.embed_content(
                model=self.model,
                contents=text
            )

            vectors.append(result.embeddings[0].values)

        return vectors

    def embed_query(self, text):
        result = self.client.models.embed_content(
            model=self.model,
            contents=text
        )

        return result.embeddings[0].values


gemini_embeddings = GeminiEmbeddings(client)

print("Gemini embedding adapter created successfully.")
````

````python
sample_text = "Machine learning is fascinating"

sample_vector = gemini_embeddings.embed_query(sample_text)

print("Input text:", sample_text)
print("Vector dimensions:", len(sample_vector))
print("First 5 values:", sample_vector[:5])
````

## 19. Save and Reuse Document Chunk Embeddings

### What Am I Doing?

I am generating embeddings for the five document chunks and saving them to a JSON file.

### Why Am I Doing This?

In the previous practical, I saved and loaded an embedding for a single sentence.

However, our RAG system has several chunks. Each chunk needs its own embedding because each chunk contains different information.

Once I have saved those embeddings, I can reuse them in later runs instead of making another Gemini request for every unchanged chunk.

### Which Tools Am I Using?

- `GeminiEmbeddings`: Generates vectors using Gemini.
- `json`: Saves the vectors to a file.
- `Path`: Defines the cache file location.
- `chunk_texts`: Contains the text of each current chunk.

### How Does It Work?

First, I collect all chunk texts. Then I check whether a compatible cache file already exists.

If the saved file matches the current text and embedding model, I load the saved vectors. Otherwise, I generate fresh embeddings and save them.

### What Should I Observe?

The number of vectors should equal the number of chunks. Each vector should have the same dimensions as the Gemini embedding model returns.


````python
import json
from pathlib import Path

# Collect the text of every current chunk
chunk_texts = [chunk.page_content for chunk in chunks]

# Use the same shared cache folder as our previous practical
cache_dir = (
    Path.cwd().parent
    / "data"
    / "shared_embedding_cache"
)
cache_dir.mkdir(parents=True, exist_ok=True)

chunk_cache_file = cache_dir / "chromadb_chunks_gemini.json"

# Check whether we can reuse previously saved embeddings
if chunk_cache_file.exists():
    cached_data = json.loads(
        chunk_cache_file.read_text(encoding="utf-8")
    )
else:
    cached_data = {}

cache_is_valid = (
    cached_data.get("model") == EMBEDDING_MODEL
    and cached_data.get("texts") == chunk_texts
    and len(cached_data.get("embeddings", [])) == len(chunk_texts)
)

if cache_is_valid:
    chunk_embeddings = cached_data["embeddings"]
    print("Loaded existing chunk embeddings from cache.")

else:
    # Generate one vector for each chunk
    chunk_embeddings = gemini_embeddings.embed_documents(
        chunk_texts
    )

    # Store the model, source texts, and vectors together
    cache_data = {
        "model": EMBEDDING_MODEL,
        "texts": chunk_texts,
        "embeddings": chunk_embeddings,
    }

    chunk_cache_file.write_text(
        json.dumps(cache_data),
        encoding="utf-8",
    )

    print("Generated and saved new chunk embeddings.")

print("Chunks:", len(chunk_texts))
print("Embeddings:", len(chunk_embeddings))
print("Vector dimensions:", len(chunk_embeddings[0]))
print("Cache file:", chunk_cache_file.resolve())
````

## 20. Initializing ChromaDB with Precomputed Gemini Embeddings

### What Am I Doing?

I am creating a local ChromaDB collection and storing the embeddings that I generated and cached in the previous step.

### Why Am I Doing This?

Gemini has already converted each document chunk into a numerical vector. I do not want to generate those same document embeddings again.

Instead, I will provide the existing vectors directly to ChromaDB, together with their text, metadata, and unique IDs.

### Which Tools Am I Using?

- `Chroma`: Connects LangChain to the ChromaDB vector store.
- `persist_directory`: Defines where the database will be saved locally.
- `upsert()`: Adds records or updates existing records with the same IDs.
- `chunk_embeddings`: Contains the previously generated Gemini vectors.

### How Does It Work?

Each vector is stored alongside the chunk it represents. Metadata records where the chunk came from.

The persistent directory allows the local collection to be reused in later sessions.

### What Should I Observe?

ChromaDB should contain five records, corresponding to the five chunks in my current dataset.


````python
# Confirm that every chunk has a corresponding embedding
assert len(chunks) == len(chunk_embeddings)
assert len(chunk_texts) == len(chunk_embeddings)

# Define a separate local ChromaDB directory
PERSIST_DIRECTORY = Path.cwd() / "chroma_db_gemini_groq_precomputed"
COLLECTION_NAME = "rag_gemini_groq_precomputed"

# Initialize ChromaDB through LangChain
vectorstore = Chroma(
    collection_name=COLLECTION_NAME,
    embedding_function=gemini_embeddings,
    persist_directory=str(PERSIST_DIRECTORY),
)

# Give every chunk a stable, unique ID
chunk_ids = [
    f"chunk_{i}"
    for i in range(len(chunks))
]

# Preserve the source metadata of every chunk
chunk_metadatas = [
    dict(chunk.metadata)
    for chunk in chunks
]

# Store the already-generated vectors directly
vectorstore._collection.upsert(
    ids=chunk_ids,
    embeddings=chunk_embeddings,
    documents=chunk_texts,
    metadatas=chunk_metadatas,
)

print("ChromaDB vector store initialized.")
print("Vectors stored:", vectorstore._collection.count())
print("Collection:", COLLECTION_NAME)
print("Database location:", PERSIST_DIRECTORY.resolve())
````

## 24. Creating the RAG Prompt Template

### What Am I Doing?

I am creating instructions that tell the language model how to answer questions using retrieved documents.

### Why Am I Doing This?

A language model can produce an answer from its existing knowledge, even when the retrieved documents do not support it.

I want to instruct the model to use the provided context and acknowledge when the answer is not available.

### Which Tool Am I Using?

`ChatPromptTemplate` from `langchain_core.prompts`.

### What Are the Two Important Variables?

- `{context}`: The relevant documents retrieved by ChromaDB.
- `{input}`: The user's question.

### What Should I Observe?

The prompt template should be created successfully. These placeholders will be filled with the actual context and question when the chain runs.


````python
from langchain_core.prompts import ChatPromptTemplate

system_prompt = """
You are an assistant for question-answering tasks.

Answer the user's question using the retrieved context below.
If the context does not contain the answer, say that you do not know
based on the provided information.
Do not invent facts.

Keep the answer concise and use a maximum of three sentences.

Retrieved context:
{context}
"""

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])

print("RAG prompt template created successfully.")
````

## 25. Building the Retrieval-Augmented Generation (RAG) Chain

### What Am I Doing?

I am connecting the ChromaDB retriever, prompt template, and Gemini LLM into one pipeline.

### Why Am I Doing This?

Without a chain, we would need to retrieve documents, format the context, create the prompt, and call the model manually for every question.

### How Does It Work?

1. The retriever finds relevant chunks.
2. The document chain inserts those chunks into the prompt.
3. Gemini generates an answer based on the retrieved context.
4. The retrieval chain returns the answer and retrieved documents.

### What Should I Observe?

The RAG chain should initialize successfully without generating an answer yet.


````python
from langchain_classic.chains import (
    create_retrieval_chain,
    create_stuff_documents_chain,
)

# Retrieve the top 3 relevant chunks
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

# Combine retrieved documents with the prompt and LLM
document_chain = create_stuff_documents_chain(
    llm,
    prompt,
)

# Connect retrieval and answer generation
rag_chain = create_retrieval_chain(
    retriever,
    document_chain,
)

print("RAG chain created successfully.")
````

### Function to Query the Modern RAG System

This function accepts a question, runs it through the RAG chain, prints the generated answer, and displays the retrieved context.

It returns the complete result so that the answer and retrieved documents can also be inspected programmatically.


````python
def query_rag_modern(question):
    print(f"Question: {question}")
    print("-" * 50)

    result = rag_chain.invoke({
        "input": question
    })

    print(f"Answer: {result['answer']}")
    print("\nRetrieved Context:")

    for i, doc in enumerate(result["context"], start=1):
        print(f"\n--- Source {i} ---")
        print(doc.page_content[:200] + "...")

        if doc.metadata:
            print("Metadata:", doc.metadata)

    return result
````

## Create RAG Chain Alternative — Using LCEL (LangChain Expression Language)

### What am I doing?

Building an alternative RAG pipeline using LangChain's composable runnable components.

### Why am I doing this?

LCEL makes the flow between retrieval, prompt formatting, model invocation, and output parsing explicit. It provides a flexible alternative to constructing the chain with `create_retrieval_chain`.


````python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import (
    RunnablePassthrough,
    RunnableParallel,
)
from langchain_core.prompts import ChatPromptTemplate
````

````python
# Create a custom prompt
custom_prompt = ChatPromptTemplate.from_template("""
Use the following context to answer the question.
If you don't know the answer based on the context, say you don't know.
Provide specific details from the context to support your answer.

Context:
{context}

Question: {question}

Answer:
""")

custom_prompt
````

````python
# Format retrieved documents
def format_docs(docs):
    return "\n\n".join(
        doc.page_content for doc in docs
    )
````

````python
# Build the LCEL RAG chain
rag_chain_lcel = (
    RunnableParallel({
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    })
    | custom_prompt
    | llm
    | StrOutputParser()
)

rag_chain_lcel
````

````python
# Test the LCEL chain
question = "What is deep learning?"

answer = rag_chain_lcel.invoke(question)

print("Question:", question)
print("\nAnswer:")
print(answer)
````

````python
for question in test_questions:
    answer = rag_chain_lcel.invoke(question)

    print(f"Question: {question}")
    print(f"Answer: {answer}")
    print("\n" + "=" * 80 + "\n")
````
